```python
import pandas as pd
import re
from io import StringIO
import matplotlib.pyplot as plt
import numpy as np
import netCDF4 as netc
import matplotlib.colors as colors
import pyproj as pyproj
from pyproj import Proj, transform
from pyproj import Transformer
import glob
import os
import sys


year=2022
spec_name = 'co2'
chemcomp = 'Koolstofdioxide'



# Read the file line-by-line to identify the problematic row
# with open(input_file_path, 'r', encoding='ISO-8859-1') as file:
    # for line_num, line in enumerate(file, start=1):
        # if line_num == 462:
            # print(f"Line {line_num}: {line}")
            # break


# Define the input and output file paths
input_file_path = 'transfer_4236_files_b57f0457/CO2_2022_companies.txt'
output_file_path = 'transfer_4236_files_b57f0457/ERI_co2_2022_metemk.csv'

# Read the input file
with open(input_file_path, 'r', encoding='ISO-8859-1') as file:
    content = file.readlines()

# Join the lines into a single string
data = ''.join(content)

# Step 1: Replace commas within quoted text with periods
data = re.sub(r'\"(.*?)\"', lambda m: m.group(0).replace(",", "."), data)

# Step 2: Replace semicolons within specific fields with commas (e.g., between "Vervaardiging van frisdranken" and "productie van mineraalwater en overig gebotteld water")
data = re.sub(r'(Vervaardiging van frisdranken);( productie van mineraalwater en overig gebotteld water)', r'\1,\2', data)

# Step 3: Remove quotation marks around values
data = re.sub(r'\"(.*?)\"', r'\1', data)

# Use StringIO to simulate a file for pandas
data_io = StringIO(data)

# Read the modified data into a DataFrame, skipping bad lines
df = pd.read_csv(data_io, sep=";", quotechar='"', engine="python", on_bad_lines="skip")

# Columns with numeric values where commas should be replaced with periods
value_columns = [
    'HOOGTE', 'LENGTE', 'BREEDTE', 'HOEK', 
    'UITSTROOMOPENING_M2', 'TEMPERATUUR', 
    'UITTREEDSNELHEID', 'VOLUMESTROOM', 
    'WARMTEINHOUD', 'EMISSIE_KG'
]

# Replace commas with periods in the specified columns
for col in value_columns:
    # Convert the column to string, replace commas with periods, then convert back to float
    df[col] = df[col].astype(str).str.replace(',', '.').astype(float)

# Verify the change by inspecting the first few rows
print(df.head())


# Save the modified DataFrame to a CSV file using ';' as the delimiter
#df.to_csv(output_file_path, sep=';', index=False)

```

```python
#To add the HARMONIE coordinates to point source data file:
#Input:

#from emission_preparation_setting import point_source_rd_file, datadir_point_source_plume_processing, year, chemcomp, spec_name

proj_RD   = pyproj.Proj('+proj=sterea +lat_0=52.15616055555555 +lon_0=5.38763888888889 +k=0.9999079 +x_0=155000 +y_0=463000 +ellps=bessel +towgs84=565.417,50.3319,465.552,-0.398957,0.343988,-1.8774,4.0725 +units=m +no_defs')
proj_HARM = pyproj.Proj('+proj=lcc +lat_1=52.500000 +lat_2=52.500000 +lat_0=52.500000 +lon_0=.000000 +k_0=1.0 +x_0=649536.512574 +y_0=1032883.739533 +a=6371220.000000 +b=6371220.000000')

#datadir = datadir_point_source_plume_processing

idx_x_emiss=df.columns.get_loc("XCO_EMISSIEPUNT")
idx_y_emiss=df.columns.get_loc("YCO_EMISSIEPUNT")
idx_x_bedrijf=df.columns.get_loc("XCO_BEDRIJF")
idx_y_bedrijf=df.columns.get_loc("YCO_BEDRIJF")

ff=df.to_numpy()
x_emiss=ff[:,idx_x_emiss]
y_emiss=ff[:,idx_y_emiss]
x_bedrijf=ff[:,idx_x_bedrijf]
y_bedrijf=ff[:,idx_y_bedrijf]


xemissHARM=np.zeros(np.size(x_emiss))
yemissHARM=np.zeros(np.size(y_emiss))
xbedrijfHARM=np.zeros(np.size(x_bedrijf))
ybedrijfHARM=np.zeros(np.size(y_bedrijf))


transformer = Transformer.from_proj(proj_RD, proj_HARM)
xemissHARM, yemissHARM = transformer.transform(x_emiss, y_emiss)
xbedrijfHARM,ybedrijfHARM=transformer.transform(x_bedrijf, y_bedrijf)

# Round coordinates to the nearest 5 meters
xemissHARM = np.round(xemissHARM / 5) * 5
yemissHARM = np.round(yemissHARM / 5) * 5
xbedrijfHARM = np.round(xbedrijfHARM / 5) * 5
ybedrijfHARM = np.round(ybedrijfHARM / 5) * 5

xemissHARM = xemissHARM.tolist()
yemissHARM = yemissHARM.tolist()
xbedrijfHARM = xbedrijfHARM.tolist()
ybedrijfHARM = ybedrijfHARM.tolist()

# Using '' as the column name
# and equating it to the list
df['XCO_EMISSIEPUNT_HARM'] = xemissHARM
df['YCO_EMISSIEPUNT_HARM'] = yemissHARM
df['XCO_BEDRIJF_HARM'] = xbedrijfHARM
df['YCO_BEDRIJF_HARM'] = ybedrijfHARM

#extract data for the needed chemical compound (specified above):

df=df.loc[df['STOF'] == chemcomp]

#specific options if you need to rename some columns before saving:

df=df.rename(columns = {'EMISSSIEOORZAAK':'EMISSIEOORZAAK'})
df=df.rename(columns = {'EMISSIE_KG':'EMISSIE'})
```

```python
#For mapped area emiss:

import pandas as pd

# Define the file paths
file_path1 = f'transfer_4236_files_b57f0457/{spec_name.upper()}_{year}_collective_emissions_aviation.txt'
file_path2 = f'transfer_4236_files_b57f0457/{spec_name.upper()}_{year}_collective_emissions_wastewatertreatment.txt'
file_path3 = f'transfer_4236_files_b57f0457/{spec_name.upper()}_{year}_collective_emissions.txt'

# Read the input files into DataFrames
df1 = pd.read_csv(file_path1, sep=";", quotechar='"', engine="python", encoding='ISO-8859-1', on_bad_lines="skip")
df2 = pd.read_csv(file_path2, sep=";", quotechar='"', engine="python", encoding='ISO-8859-1', on_bad_lines="skip")
df3 = pd.read_csv(file_path3, sep=";", quotechar='"', engine="python", encoding='ISO-8859-1', on_bad_lines="skip")

# Ensure that 'EMISSIE_KG' is properly formatted by replacing commas with dots and removing whitespace
for df in [df1, df2, df3]:
    if 'EMISSIE_KG' in df.columns:
        df['EMISSIE_KG'] = df['EMISSIE_KG'].str.replace(',', '.', regex=False).str.strip()  # Replace ',' with '.' and strip whitespace
        df['EMISSIE_KG'] = pd.to_numeric(df['EMISSIE_KG'], errors='coerce')  # Convert to float, coerce errors to NaN


# Merge the DataFrames
merged_df = pd.concat([df1, df2, df3], ignore_index=True)

# Drop duplicate rows
merged_df = merged_df.drop_duplicates()


# Display the first few rows and the columns of the dataframe
print(merged_df.head())
print(merged_df.columns)

#save data into the file that can be further used to prepare the point source input to the model:
merged_df.to_csv('ER_map_'+str(year)+'_'+str(spec_name)+'.csv',sep=';', encoding='ISO-8859-1') 
```
