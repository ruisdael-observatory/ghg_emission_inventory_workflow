```python
#This program useful for coordinates convertion between HARM, RD, and geographical lat long:
###########################################################################
#Author: Arseni Doyennel
#e-mail: a.doyennel@vu.nl
#Date: 17/05/2023
#Place: VU Amsterdam

%matplotlib inline

import matplotlib.pyplot as plt
import numpy as np
import netCDF4 as netc
import pandas as pd
import matplotlib.colors as colors
import pyproj as pyproj
from pyproj import Proj, transform
from pyproj import Transformer
import glob
import math
import os
import sys
sys.path.insert(1,'/Users/darseni/Documents/emissions_prep_final/area_emissions_main/')
from emission_preparation_setting import x0, y0, xn, yn, xres, yres


proj_RD   = pyproj.Proj('+proj=sterea +lat_0=52.15616055555555 +lon_0=5.38763888888889 +k=0.9999079 +x_0=155000 +y_0=463000 +ellps=bessel +towgs84=565.417,50.3319,465.552,-0.398957,0.343988,-1.8774,4.0725 +units=m +no_defs')
proj_HARM = pyproj.Proj('+proj=lcc +lat_1=52.500000 +lat_2=52.500000 +lat_0=52.500000 +lon_0=.000000 +k_0=1.0 +x_0=649536.512574 +y_0=1032883.739533 +a=6371220.000000 +b=6371220.000000')


datadir = '/Users/darseni/Documents/emissions_prep_final/emission_raw_data/emission_files/'


xmax=x0 + int(xn * xres) #upper right corner X-coord of the domain in HARMONIE coordinates
ymax=y0 + int(yn * yres) #upper right corner Y-coord of the domain in HARMONIE coordinates 

```

```python
#from HARMONIE to RD-coordinates:

transformer = Transformer.from_proj(proj_HARM, proj_RD)
x0_RD_0, y0_RD_0 = transformer.transform(x0, y0)
x0_RD_max, y0_RD_max = transformer.transform(xmax, ymax)

#from RD-coordinates to GEO lat and long:

transformer = Transformer.from_crs("EPSG:28992", "WGS84")
lat_0, lon_0 = transformer.transform(x0_RD_0, y0_RD_0)
lat_max, lon_max = transformer.transform(x0_RD_max, y0_RD_max)
```

```python
#LAT and LONG of the domain center (need to put in namoptions for DALES running):

lat_center = ((lat_max - lat_0)/2)+lat_0
lon_center = ((lon_max - lon_0)/2)+lon_0

print("latitude of domain center:", lat_center, "Longitude of domain center:", lon_center)
```

```python
#if needed a convertion from GEO lat and long to HARMONIE coordinates:

#The coarse domain, resolution 
#50.6 - 53.9 N   3.3 deg = 366 km
#1.8 - 9.1 E     7.3 deg = 496.7 km



lat_0 = 50
lon_0 = 2

lat_max = 54
lon_max = 9.2

#from GEO lat and long to RD-coordinates:

transformer = Transformer.from_crs("WGS84", "EPSG:28992")
x0_RD_0, y0_RD_0 = transformer.transform(lat_0, lon_0)
x0_RD_max, y0_RD_max = transformer.transform(lat_max, lon_max)

#from RD-coordinates to HARMONIE coordinates:

transformer = Transformer.from_proj(proj_RD, proj_HARM)
x0_back, y0_back = transformer.transform(x0_RD_0, y0_RD_0)
xmax_back, ymax_back = transformer.transform(x0_RD_max, y0_RD_max)
```

```python
xres = 200
yres = 200


x0_round=(int(math.ceil(x0_back / xres)) * xres)-xres
y0_round=(int(math.ceil(y0_back / yres)) * yres)-yres
xmax_round=(int(math.ceil(xmax_back / xres)) * xres)+xres
ymax_round=(int(math.ceil(ymax_back / yres)) * yres)+yres
```

```python
print("Min X HARM:", x0_back, "Min Y HARM:", y0_back,"Max X HARM:", xmax_back, "Max Y HARM:", ymax_back)

print("rounded*", "Min X HARM:",x0_round, "Min Y HARM:", y0_round,"Max X HARM:", xmax_round, "Max Y HARM:", ymax_round)
```

```python
#from HARMONIE to RD-coordinates:

transformer = Transformer.from_proj(proj_HARM, proj_RD)
x0_RD_0, y0_RD_0 = transformer.transform(x0_round, y0_round)
x0_RD_max, y0_RD_max = transformer.transform(xmax_round, ymax_round)

#from RD-coordinates to GEO lat and long:

transformer = Transformer.from_crs("EPSG:28992", "WGS84")
lat_0, lon_0 = transformer.transform(x0_RD_0, y0_RD_0)
lat_max, lon_max = transformer.transform(x0_RD_max, y0_RD_max)
```

```python
print("from rounded coord.", "Min LAT:", lat_0, "Min LON:", lon_0,"Max LAT:", lat_max, "Max LON:", lon_max)
```

```python
xn = (xmax_round-x0_round)/xres
yn = (ymax_round-y0_round)/yres

print("from rounded coord.", "xn:", xn, "yn:", yn)
```
