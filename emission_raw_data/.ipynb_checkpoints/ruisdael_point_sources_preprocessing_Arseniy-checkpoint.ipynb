```python
def new_proj_points(df, proj_old, proj_new, old_suffix, new_suffix):
    def transf_proj(x,y):
        transformer = Transformer.from_proj(proj_old,proj_new)
        return transformer.transform(x,y)
    
    df[f"EMISSION_LOCATION_X_{new_suffix}"], df[f"EMISSION_LOCATION_Y_{new_suffix}"] = np.vectorize(transf_proj)(df['EMISSION_LOCATION_X'], df['EMISSION_LOCATION_Y'])
    df = df.rename(columns={"EMISSION_LOCATION_X": f"EMISSION_LOCATION_X_{old_suffix}", "EMISSION_LOCATION_Y": f"EMISSION_LOCATION_Y_{old_suffix}"})
    return df
```

```python
for f in glob.glob(datadir+'point_sources_with_snap_'+str(year)+'_HARM.csv'):
    os.remove(f)


df_uni      = unify_points(df_orig)
df_uni_harm = new_proj_points(df_uni, proj_RD, proj_HARM, 'RD', 'HARM')
df_uni_harm.to_csv(datadir+'point_sources_with_snap_'+str(year)+'_HARM.csv', encoding='utf-8')
```
