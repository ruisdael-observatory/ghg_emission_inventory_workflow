```python
#Program tro create a possible domain to simulate on snellius
###########################################################################
#Author: Arseni Doyennel
#e-mail: a.doyennel@vu.nl
#Date: 10/07/2023
#Place: VU Amsterdam

%matplotlib inline

import matplotlib.pyplot as plt
import numpy as np
import netCDF4 as netc
import pandas as pd
import matplotlib.colors as colors
import pyproj as pyproj
from pyproj import Proj, transform
from pyproj import Transformer
import glob
import math
import os

proj_RD   = pyproj.Proj('+proj=sterea +lat_0=52.15616055555555 +lon_0=5.38763888888889 +k=0.9999079 +x_0=155000 +y_0=463000 +ellps=bessel +towgs84=565.417,50.3319,465.552,-0.398957,0.343988,-1.8774,4.0725 +units=m +no_defs')
proj_HARM = pyproj.Proj('+proj=lcc +lat_1=52.500000 +lat_2=52.500000 +lat_0=52.500000 +lon_0=.000000 +k_0=1.0 +x_0=649536.512574 +y_0=1032883.739533 +a=6371220.000000 +b=6371220.000000')
```

```python
#set the resolution of simulation:

#on snellius each node = 128 tasks
#nprocx and nprocy nust be a devidors of xn and yn and nprocx*nprocy should be a devisor of 128
# so we need to find a proper domain corners (within the area of interest) to fit to this setting:


#Chosen domain (Ransdad simulations):

#x0 and y0 (left lower corner of the domain) can be chosen as you wish but less than large domain (for open BCs):

#52.4-52.5 N and 4.5-4.7 E

x0 = 935000 
y0 = 1015000   

#from HARMONIE to RD-coordinates:
transformer = Transformer.from_proj(proj_HARM, proj_RD)
x0_RD_0, y0_RD_0 = transformer.transform(x0, y0)

#from RD-coordinates to GEO lat and long:
transformer = Transformer.from_crs("EPSG:28992", "WGS84")
lat_0, lon_0 = transformer.transform(x0_RD_0, y0_RD_0)

print("Min lat and lon.", "LAT:", lat_0, "LON:", lon_0)
```

```python
#domain settings:
####################################################

xres = 50
yres = 50

xn = 1024 #should be equally devided by nprocx
yn = 1024 #should be equally devided by nprocy

nprocx = 16 #should be an equal devider of xn but that the result will be /2:   16,32,64,128...
nprocy = 16 #should be an equal devider of yn but that the result will be /2:   16,32,64,128...

#xn and yn should be equally devided by a number of processes (crusial!!!):

if (xn % nprocx == 0) and (yn % nprocy == 0):
        print('x0/y0 equally devided by nprocx/nprocy!')
else:
        raise Warning("change xn and yn to those equally devided by nprocx/nprocy!")
        
if (xn/nprocx <= 128) and (yn/nprocy <= 128):
        print('xn/nprocx and yn/nprocy less than 128!')
else:
        raise Warning("xn/nprocx and yn/nprocy more than 128 -> increase nproc or decrease the domain size!")
        
print('xn/nprocx:',xn/nprocx,'yn/nprocy:',yn/nprocy) #should be less than 128 (1 node)


xmax=x0 + int(xn * xres) #upper right corner X-coord of the domain in HARMONIE coordinates
ymax=y0 + int(yn * yres) #upper right corner Y-coord of the domain in HARMONIE coordinates 


#from HARMONIE to RD-coordinates:
transformer = Transformer.from_proj(proj_HARM, proj_RD)
x0_RD_0, y0_RD_0 = transformer.transform(xmax, ymax)

#from RD-coordinates to GEO lat and long:
transformer = Transformer.from_crs("EPSG:28992", "WGS84")
lat_max, lon_max = transformer.transform(x0_RD_0, y0_RD_0)

print("Max lat and lon.", "LAT:", lat_max, "LON:", lon_max)
```
